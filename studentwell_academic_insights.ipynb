```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("STUDENTWELL - ACADEMIC INSIGHTS")
print("Sleep, Screen Time and Academic Performance")

# Load dataset
df = pd.read_csv("studentwell_dataset (3).csv")
df.columns = df.columns.str.strip()

print("\n1. DATASET INFORMATION")
print("Total students:", df.shape[0])
print("Total columns:", df.shape[1])

print("\n2. FIRST FIVE RECORDS")
display(df.head())

print("\n3. COLUMN NAMES")
print(df.columns.tolist())

print("\n4. MISSING VALUES")
display(df.isnull().sum())

print("\n5. DUPLICATE ROWS")
print("Duplicates:", df.duplicated().sum())

# Analyze available numerical columns
features = [
    "avg_sleep_hours_weekday",
    "daily_screen_time_hours",
    "cgpa",
    "weekly_study_hours",
    "class_attendance_percent"
]

for col in features:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

available = [col for col in features if col in df.columns]

print("\n6. SUMMARY STATISTICS")
if available:
    display(df[available].describe().round(2))
else:
    print("Expected numerical columns were not found.")

print("\n7. GRAPHS")

for col, title in [
    ("avg_sleep_hours_weekday", "Sleep Hours Distribution"),
    ("daily_screen_time_hours", "Daily Screen Time Distribution"),
    ("cgpa", "CGPA Distribution")
]:
    if col in df.columns and df[col].notna().any():
        plt.figure(figsize=(7, 4))
        plt.hist(df[col].dropna(), bins=15, edgecolor="black")
        plt.title(title)
        plt.xlabel(col)
        plt.ylabel("Number of Students")
        plt.tight_layout()
        plt.show()

print("\n8. RELATIONSHIP ANALYSIS")

for x, y, title in [
    ("daily_screen_time_hours", "cgpa", "Screen Time vs CGPA"),
    ("avg_sleep_hours_weekday", "cgpa", "Sleep Hours vs CGPA")
]:
    if x in df.columns and y in df.columns:
        data = df[[x, y]].dropna()

        if not data.empty:
            plt.figure(figsize=(7, 4))
            plt.scatter(data[x], data[y])
            plt.title(title)
            plt.xlabel(x)
            plt.ylabel(y)
            plt.tight_layout()
            plt.show()

            print(title, "correlation:",
                  round(data[x].corr(data[y]), 3))

print("\n9. CORRELATION MATRIX")

numeric_df = df.select_dtypes(include=np.number)
if numeric_df.shape[1] >= 2:
    corr = numeric_df.corr()
    display(corr.round(2))

    plt.figure(figsize=(10, 7))
    plt.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)
    plt.colorbar(label="Correlation")
    plt.xticks(range(len(corr.columns)),
               corr.columns, rotation=90)
    plt.yticks(range(len(corr.columns)),
               corr.columns)
    plt.title("Student Data Correlation Heatmap")
    plt.tight_layout()
    plt.show()

print("\n10. PROJECT CONCLUSION")
print("The dataset was loaded and analyzed successfully.")
print("Tables, statistics, and graphs were generated.")
print("Correlation shows association, not causation.")
print("Analysis completed!")
```
